```python
pin_district_count = (
    df_WestBengal.groupby('pincode')['district_clean']
      .nunique()
      .reset_index(name='district_count')
)
```

```python
problem_pins = pin_district_count[
    pin_district_count['district_count'] > 1
]
```

```python
df_flagg = df_WestBengal.merge(
    problem_pins[['pincode']],
    on='pincode',
    how='inner'
)
```

```python
df_flagg
```

```python
flagged_pincode_dominant=df_flagg.groupby(['district_clean', 'pincode'])[['demo_age_5_17','demo_age_17_']].sum().reset_index()
```
